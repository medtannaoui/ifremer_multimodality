les données micro wave (passive microwave)

```python
annee = annees[-5] #choisir une année
bassin = bassins[2] #choisir un bassin
cyn_number = '01' #choisir un numéro de cyclone
# for i,j in enumerate(os.listdir(f"{main_path}/{annee}/{bassin}/{cyn_number}")):
#     print(i,j)
if True : 
    file_path = os.listdir(f"{main_path}/{annee}/{bassin}/{cyn_number}")[0] #choisir un fichier netcdf
    f = nc.Dataset(f"{main_path}/{annee}/{bassin}/{cyn_number}/{file_path}") #ouvrir le fichier netcdf avec netCDF4
    groupes = f.groups.keys() #obtenir les groupes du fichier netcdf
    ds_pw = xr.open_dataset(f"{main_path}/{annee}/{bassin}/{cyn_number}/{file_path}",group=f"{list(groupes)[-3]}/S3")  #ouvrir le groupe 'PW/S3' avec xarray

    print(list(ds_pw.data_vars.keys()))
    var = list(ds_pw.data_vars.keys())[6] #selectionner la variable à afficher (passive_microwave)
    plt.figure(figsize=(8, 6))
    plt.pcolormesh(ds_pw["x"], ds_pw["y"], ds_pw[var], cmap="turbo")  # 'turbo' = bonne palette thermique
    plt.colorbar(label=f"{var} (K)")
    plt.title(f"{var} - {ds_pw.attrs['instrument']} {ds_pw.attrs['platform']}")
    plt.xlabel("x (km)")
    plt.ylabel("y (km)")
    plt.axis("equal")
    plt.show()
```
