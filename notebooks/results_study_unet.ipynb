```python
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np

df = pd.read_csv("/scale/user/mtannaou/alternance/src/IR_to_SAR/ML_IR_SAR/combined_loss/results.csv") 
```

```python
import numpy as np
import matplotlib.pyplot as plt


cols_vmax = [
    "rmse_vmax_global",
    "rmse_vmax_cat_1",
    "rmse_vmax_cat_2",
    "rmse_vmax_cat_3",
    "RMSE_ANGGREK_Vmax"
]

categories = ["Global", "Cat1", "Cat2", "Cat3","Anggrek"]

n_exp = len(df)
x = np.arange(len(categories))
width = 0.8 / n_exp

plt.figure(figsize=(12,5))

for i in range(n_exp):
    values = df.loc[i, cols_vmax].values
    plt.bar(x + i*width, values, width, label=f"Experiment_{i+1}")

plt.xticks(x + width*(n_exp - 1)/2, categories)
plt.ylabel("RMSE Vmax")
plt.title("RMSE Vmax by category")
plt.legend()
plt.tight_layout()
plt.grid()
plt.show()
```

```python
cols_rmax = [
    "rmse_rmax_global",
    "rmse_rmax_cat_1",
    "rmse_rmax_cat_2",
    "rmse_rmax_cat_3",
]
categories = ["Global", "Cat1", "Cat2", "Cat3"]
n_exp = len(df)
x = np.arange(len(categories))
width = 0.8 / n_exp
plt.figure(figsize=(12,5))

for i in range(n_exp):
    values = df.loc[i, cols_rmax].values
    plt.bar(x + i*width, values, width, label=f"Experiment_{i+1}")

plt.xticks(x + width*(n_exp - 1)/2, categories)
plt.ylabel("RMSE Rmax")
plt.title("RMSE Rmax by category")
plt.legend()
plt.tight_layout()
plt.grid()
plt.show()
```

## Loss Used comparison

```python
import numpy as np
import matplotlib.pyplot as plt

df_loss = df.iloc[[0,3,8]]
cols_vmax = [
    "rmse_vmax_global",
    "rmse_vmax_cat_1",
    "rmse_vmax_cat_2",
    "rmse_vmax_cat_3",
    "RMSE_ANGGREK_Vmax"
]

categories = ["Global", "Cat1", "Cat2", "Cat3","Anggrek"]

n_exp = len(df_loss)
x = np.arange(len(categories))
width = 0.8 / n_exp

plt.figure(figsize=(16,5))

for i in range(n_exp):
    values = df_loss.iloc[i][cols_vmax].values
    plt.bar(
        x + i*width,
        values,
        width,
        label=df_loss.iloc[i]["Loss Used"]
    )

plt.xticks(x + width*(n_exp - 1)/2, categories)
plt.ylabel("RMSE Vmax")
plt.title("RMSE Vmax by category")
plt.legend()
plt.tight_layout()
plt.grid()
plt.show()
```

```python
cols_rmax = [
    "rmse_rmax_global",
    "rmse_rmax_cat_1",
    "rmse_rmax_cat_2",
    "rmse_rmax_cat_3",
]
categories = ["Global", "Cat1", "Cat2", "Cat3"]
n_exp = len(df_loss)
x = np.arange(len(categories))
width = 0.8 / n_exp
plt.figure(figsize=(16,5))

for i in range(n_exp):
    values = df_loss.iloc[i][cols_rmax].values
    plt.bar(x + i*width, values, width, label=df_loss["Loss Used"].unique()[i])

plt.xticks(x + width*(n_exp - 1)/2, categories)
plt.ylabel("RMSE Rmax")
plt.title("RMSE Rmax by category")
plt.legend()
plt.tight_layout()
plt.grid()
plt.show()
```
